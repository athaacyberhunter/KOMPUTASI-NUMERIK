# CELL 2: PROSES ELIMINASI MAJU (MEMBENTUK MATRIKS SEGITIGA ATAS)
# ==============================================================================
def eliminasi_maju(matriks):
    # Membuat salinan matriks agar data asli di Cell 1 tidak berubah
    M = matriks.copy()
    # Mengambil jumlah baris (n = 3)
    n = len(M)
    
    # Loop utama untuk menentukan kolom pivot (i = 0 dan i = 1)
    for i in range(n - 1):
        # Partial Pivoting: Jika elemen diagonal utama (pivot) bernilai 0,
        # tukar baris saat ini dengan baris di bawahnya yang tidak nol
        if M[i][i] == 0:
            for p in range(i + 1, n):
                if M[p][i] != 0:
                    M[[i, p]] = M[[p, i]]  # Menukar posisi baris i dan baris p
                    break
        
        # Loop untuk mengeliminasi (meng-nol-kan) elemen di bawah pivot pada kolom i
        for k in range(i + 1, n):
            # Menghitung rasio/faktor pengali antara baris target (k) dan baris pivot (i)
            faktor = M[k][i] / M[i][i]
            
            # Melakukan Operasi Baris Elementer (OBE): Baris_k = Baris_k - (faktor * Baris_i)
            for j in range(i, n + 1):
                M[k][j] = M[k][j] - faktor * M[i][j]
                
    # Mengembalikan matriks yang sudah berbentuk segitiga atas
    return M

# Menjalankan fungsi eliminasi maju pada matriks A dan mencetak hasilnya

# CELL 3: PROSES SUBSTITUSI BALIK (MENCARI NILAI X, Y, Z)
# ==============================================================================
def substitusi_balik(M):
    # Mengambil jumlah baris (n = 3)
    n = len(M)
    
    # Pengecekan apakah elemen pivot terakhir bernilai nol (solusi tidak unik/tak hingga)
    if M[n - 1][n - 1] == 0:
        raise ValueError("Sistem tidak memiliki solusi unik (matriks singular).")
    
    # Membuat array kosong berukuran 3 untuk menampung hasil [x, y, z]
    solusi = np.zeros(n)
    
    # Menghitung variabel dari baris paling bawah (indeks 2) naik ke baris atas (indeks 0)
    for i in range(n - 1, -1, -1):
        # Menghitung jumlah perkalian koefisien dengan variabel yang sudah ditemukan sebelumnya
        jumlah_diketahui = 0.0
        for j in range(i + 1, n):
            jumlah_diketahui += M[i][j] * solusi[j]
            
        # Rumus substitusi balik: x_i = (konstanta - jumlah_diketahui) / koefisien_diagonal
        solusi[i] = (M[i][n] - jumlah_diketahui) / M[i][i]
        
    # Mengembalikan array solusi [x, y, z]
    return solusi
