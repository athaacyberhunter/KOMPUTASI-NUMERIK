import numpy as np

print("Library NumPy berhasil dimuat!")

# Contoh SPL 3 Variabel:
# 2x +  y -  z =  8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

# Matriks koefisien (A) dan vektor konstanta (B) digabung menjadi Matriks Augmented (M)
M = np.array([
    [ 2.0,  1.0, -1.0,   8.0],
    [-3.0, -1.0,  2.0, -11.0],
    [-2.0,  1.0,  2.0,  -3.0]
], dtype=float)

print("Matriks Augmented Awal (M):")
print(M)

n = len(M)

for i in range(n):
    # Partial Pivoting: Tukar baris jika elemen diagonal utama bernilai 0
    if M[i][i] == 0.0:
        for k in range(i + 1, n):
            if M[k][i] != 0.0:
                M[[i, k]] = M[[k, i]]
                break
                
    # Eliminasi elemen di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j][i] / M[i][i]
        M[j] = M[j] - faktor * M[i]

print("Matriks Segitiga Atas setelah Eliminasi Maju:")
print(np.round(M, 4))

# Cek apakah sistem memiliki solusi unik
if M[2][2] == 0:
    if M[2][3] != 0:
        print("SPL tidak memiliki solusi (Inkonsisten).")
    else:
        print("SPL memiliki tak hingga banyak solusi.")
else:
    # Hitung z, y, dan x secara mundur
    z = M[2][3] / M[2][2]
    y = (M[1][3] - M[1][2] * z) / M[1][1]
    x = (M[0][3] - M[0][1] * y - M[0][2] * z) / M[0][0]

    print("=== HASIL PENYELESAIAN SPL ===")
    print(f"x = {x:.4f}")
    print(f"y = {y:.4f}")
    print(f"z = {z:.4f}")
